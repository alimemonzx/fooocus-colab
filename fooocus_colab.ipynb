## How to start
1. **Runtime → Change runtime type → T4 GPU** (the free GPU works).
2. Click ▶ on the cell below. If Colab warns that the notebook wasn't authored by Google, click **Run anyway**.
3. Wait about 8–10 minutes while it installs and downloads models.
4. Open the `https://….gradio.live` link printed at the end. Keep this tab open while you use Fooocus.

Everything runs in **your own** Colab account. To keep your own editable copy, use **File → Save a copy in Drive**. Colab deletes everything when the session ends, so download any images you want to keep.
